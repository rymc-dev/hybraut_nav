# Hybraut Model Demo

This demo shows examples of the hybrid automaton, It will show how 

```python
from hybraut_factory.amdl import HybridAutomatonFactory
from hybraut_models import HybridAutomaton
import yaml


tb3_amdl_example_path = "/home/ryan/ros2_ws/src/hybraut_ros2/example_amdls/hybraut_tb3.amdl.yml"
with open(tb3_amdl_example_path, "r") as f:
    tb3_amdl_dict = yaml.safe_load(f)
    
    
tb3_hybraut_model: HybridAutomaton = HybridAutomatonFactory.register_automaton(tb3_amdl_dict)

print (str(tb3_hybraut_model))
print (repr(tb3_hybraut_model))
```

Diagram generation of the automaton

```python
from hybraut_diag_gen import HybrautDiag
import os

HybrautDiag(output_directory=os.path.join(os.getcwd(), "assets")).generate_mermaid_diagrams(
    automaton_model=tb3_hybraut_model,
)
```

```python
from hybraut_execution_engine import ExecutionEngine
import rclpy
from rclpy.node import Node
from rclpy.executors import MultiThreadedExecutor
import threading

rclpy.init()

executor = MultiThreadedExecutor()
node = Node('node')
executor.add_node(node)

thread = threading.ThreadError(target=executor.spin, daemon=True)
thread.start()

hybraut_execution_engine: ExecutionEngine = ExecutionEngine(node=node, automaton_model=tb3_hybraut_model)
hybraut_execution_engine.activate()

import time
# sleep the main thread
time.sleep(30.0)


rclpy.shutdown()

```
